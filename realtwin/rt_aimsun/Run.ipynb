```python
#Step 5.1 Generate Scenario
import re

cmd = [
    aconsole_path,
    "-script",
    step_51_script_path,
    model_path,
]

process = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                           text=True, encoding="utf-8", errors="replace")
out, _ = process.communicate()
print(out)

# Capture the created replication's Aimsun ID from the output for the simulation step
m = re.search(r"REPLICATION_ID=(\d+)", out)
target_id = m.group(1) if m else None
print("Replication Aimsun ID (target_id):", target_id)
```

```python
# Run the simulation for test
# target_id comes from the Step 5.1 cell above (the created replication's Aimsun ID)

if not target_id:
    raise ValueError("target_id is not set - run the Step 5.1 cell first")

cmd = [
    aconsole_path,
    "--project",
    model_path,
    "--command",
    "execute",
    "--target",
    target_id,
]

process = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                           text=True, encoding="utf-8", errors="replace")
out, _ = process.communicate()
print(out)
```
