```python
import pandas as pd
import csv
import requests
import xml.etree.ElementTree as ET
from math import atan, exp, cos
from pyproj import Transformer
```

```python
def vissim_coords_to_lat_long(xVissim, yVissim):
    # variables from vissim net settings
    xRefNet = 0.000
    yRefNet = 0.000
    # xRefMap = -9485531.762
    # yRefMap = 4165877.412
    #Shallowford
    xRefMap = -9468636.559
    yRefMap = 4164801.456
    # constant values
    PI = 3.14159265358979
    EarthRadius = 6378137
    CorrectionFactorMercator = 1.0011202320000001

    # deriving vissim local scal factor
    LatitudeRefPointMap = ( 2 * atan( exp( CorrectionFactorMercator * yRefMap / EarthRadius ) ) - PI / 2 ) / ( PI / 180 )
    LocalScaleFactor = 1 / cos( LatitudeRefPointMap * PI / 180 )

# # vissim coordinates to be transformed
# xVissim = -308.909
# yVissim = -481.288

    # calculating xy coordinates in Mercator
    xMercator = ( xVissim - xRefNet ) * LocalScaleFactor + xRefMap
    yMercator = ( yVissim - yRefNet ) * LocalScaleFactor + yRefMap

    # transform Mercator coordinates to WGS84
    merc2wgs84 = Transformer.from_crs('ESRI:53004', 'EPSG:4326')
    wgs84 = merc2wgs84.transform(xMercator, yMercator)
    print(wgs84)
    p = [wgs84[0], wgs84[1]]
    print (p)
    return (p)
```

```python
# function to get divergence link id list
def get_vissim_div_link_ids():
    file_name = 'C:\\Users\\ets\\Desktop\\RealTwins-Abhilasha\\sc_test\\links_divergence_inventory.csv'
    df = pd.read_csv (file_name)
    rslt_df = df[df['divergence'] == "divergence"]
    list_div_link_ids = rslt_df['Link_No'].unique().tolist()
    print (list_div_link_ids)
    return list_div_link_ids
```

```python
# function to get divergence link coordinates in lat long 
def get_link_coords(div_link_id):
    file_name = 'C:\\Users\\ets\\Desktop\\RealTwins-Abhilasha\\sc_test\\links_divergence_inventory.csv'
    df = pd.read_csv (file_name)
    Vissim_X = df.loc[df['Link_No'] == div_link_id, 'StartLinkCoord-X'].iloc[0]
    Vissim_Y = df.loc[df['Link_No'] == div_link_id, 'StartLinkCoord-X'].iloc[0]
    lat_long = vissim_coords_to_lat_long(Vissim_X, Vissim_Y)
    return lat_long
```
