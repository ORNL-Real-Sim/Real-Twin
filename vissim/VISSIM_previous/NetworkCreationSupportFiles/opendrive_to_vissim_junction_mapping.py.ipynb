```python
#function to convert opendrive coordinates to lat/long
from pyproj import Proj, transform

# from pyproj import Transformer

def opendrive_coords_to_lat_long(x,y):
    # offset of the opendrive file
    # xodr_x0=-639371.48
    # xodr_y0=-3848823.33
    ##Shallowford
    xodr_x0 = -667733.87
    xodr_y0=-3878705.38

    # initialize projection objects
    wgs84Proj = Proj('EPSG:4326')
    xodrProj = Proj('EPSG:32616')
    
#     transformer = Transformer.from_crs("EPSG:4326", "EPSG:3857")
#     transformer.transform(12, 12)

    
    # do the projection
    wgs84_x0, wgs84_y0 = transform(xodrProj, wgs84Proj, x-xodr_x0, y-xodr_y0)
    print('xodr/sumo coordinates {},{}, wgs84 coordinates {},{}'.format(x, y, wgs84_x0, wgs84_y0))
    fin_list = [x, y, wgs84_x0, wgs84_y0]
    return fin_list
```

```python
from math import radians, cos, sin, asin, sqrt

def haversine(lon1, lat1, lon2, lat2):
    """
    Calculate the great circle distance in kilometers between two points 
    on the earth (specified in decimal degrees)
    """
    # convert decimal degrees to radians 
    lon1, lat1, lon2, lat2 = map(radians, [lon1, lat1, lon2, lat2])

    # haversine formula 
    dlon = lon2 - lon1 
    dlat = lat2 - lat1 
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * asin(sqrt(a)) 
#     r = 6371 # Radius of earth in kilometers. Use 3956 for miles. Determines return value units.
    r = 3956*5280
    return c * r
```

```python
# function to get all points within a circular area - parameters - center lat long and radius
def check_points_within_radius(j_id):
    radius = get_opendrive_junc_vissim_radius(j_id)*0.1
    
    # get junction mean coordinates
    x_jun_odr = get_junction_mean_lat_long(j_id)[4]
    y_jun_odr = get_junction_mean_lat_long(j_id)[5]
    
    list_possible_elements_in_intersection = for_opendrive_junc_get_closest_vissim_div_links(j_id)
#     print (list_possible_elements_in_intersection)
    big_list = []
    for i in range(len(list_possible_elements_in_intersection)):
#         print (list_possible_elements_in_intersection[i])

        my_item = next((item for item in dictionary_vissim_all_roads if item['Link_No'] == list_possible_elements_in_intersection[i]), None)
#         print (my_item)
        
        # get opendrive lat long info for the elements id from 
        x_lat = my_item['Latitude']
        y_long = my_item['Longitude']
        
        square_dist_2 = haversine(y_jun_odr, x_jun_odr, y_long, x_lat)
        
        if square_dist_2<= radius**2:
            print (radius**2 - square_dist_2)
            big_list.append(list_possible_elements_in_intersection[i])
        else:
            print ("not within radius")
    return big_list
```
