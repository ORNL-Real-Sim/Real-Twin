```python
# create element tree object
tree = ET.parse(xml_file_name)

# print (tree)

# get root element
root = tree.getroot()
# print (root)

big_road_info_list = []
# df = pd.DataFrame(columns=['veh_id', 'veh_type', 'veh_depart_tm', 'route_edges'])
i = 0
for child in root.findall('./road'):
#     print (i)
#     print (child.tag)
#     print (child.attrib)
    small_list = [child.attrib['name'], child.attrib['length'], child.attrib['id'], child.attrib['junction']]
    preds = child.findall('./link/predecessor')
    succs = child.findall('./link/successor')
    geos = child.findall('./planView/geometry')
    
    if len(preds)>0:
        for more_items_pred in child.findall('./link/predecessor'):
            print (more_items_pred)
            small_list.append('predecessor')
            small_list.append(more_items_pred.attrib["elementType"])
            small_list.append(more_items_pred.attrib["elementId"])
    else:
        small_list.append('predecessor')
        small_list.append('None')
        small_list.append('None')
    
    if len(succs)>0:
        for more_items_succ in child.findall('./link/successor'):
            small_list.append('successor')
            small_list.append(more_items_succ.attrib["elementType"])
            small_list.append(more_items_succ.attrib["elementId"])
    else:
        small_list.append('successor')
        small_list.append('None')
        small_list.append('None')

    if len(geos)>0:
        for more_items_geo in child.findall('./planView/geometry'):
    #         print (more_items_geo.tag)
    #         small_list.append('successor')
            small_list.append(more_items_geo.attrib["s"])
            small_list.append(more_items_geo.attrib["x"])
            small_list.append(more_items_geo.attrib["y"])
            small_list.append(more_items_geo.attrib["hdg"])
            small_list.append(more_items_geo.attrib["length"])
    else:
        small_list.append("none")
        small_list.append("none")
        small_list.append("none")
        small_list.append("none")
        small_list.append("none")
    
#     print (small_list)
    small_list_cut = small_list[0:15]
#     print (small_list_cut)
    
    big_road_info_list.append(small_list_cut)
    
print (big_road_info_list)
```
