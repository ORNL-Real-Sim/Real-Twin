```python
# function to get coordinate of end of incoming road and junction - start point of successor of the incoming road to the junction
def get_coord_of_inc_road_to_junction(j_id):
     
    junc_info_fname = 'C:\\Users\\ets\\Desktop\\RealTwins-Abhilasha\\SUMO_setup_check\\chattdtwn-main_v2\\chattdtwn-main\\python_scripts\\link_mapping_scripts\\opendrive_junction_info.csv'
    df_junc_info =  pd.read_csv(junc_info_fname)
    
    road_info_fname ='C:\\Users\\ets\\Desktop\\RealTwins-Abhilasha\\SUMO_setup_check\\chattdtwn-main_v2\\chattdtwn-main\\python_scripts\\link_mapping_scripts\\opendrive_road_info_inventory.csv'
    df_road_info = pd.read_csv(road_info_fname)
    
    incoming_road_list = list_of_inc_roads_for_a_junction(j_id)
    big_list = []
    
    for i in range(len(incoming_road_list)):
        #filter junction df for incoming road and get conneting roads
        print (incoming_road_list[i])
        rslt_df_junc_info = df_junc_info[df_junc_info['incoming_road']== incoming_road_list[i]]
        connecting_rd_list = rslt_df_junc_info['connecting_road'].unique().tolist()
        x_fin = 0
        y_fin = 0
        hdg_fin = 0
        
        for j in range(len(connecting_rd_list)):
#             rslt_df_road_info = df_road_info[df_road_info['road_id']==connecting_rd_list[j]]
            x_coord = df_road_info.loc[df_road_info['road_id'] == connecting_rd_list[j], 'road_start_point_coordinate_X']
            print (x_coord)
            y_coord = df_road_info.loc[df_road_info['road_id'] == connecting_rd_list[j], 'road_start_point_coordinate_Y']
            hdg_coord = df_road_info.loc[df_road_info['road_id'] == connecting_rd_list[j], 'road_start_point_coordinate_hdg']
            x_fin = x_fin+float(x_coord)
            y_fin = y_fin+float(y_coord)
            hdg_fin = hdg_fin+float(hdg_coord)
            
        x_mean = float(x_fin)/len(connecting_rd_list)
        y_mean = y_fin/len(connecting_rd_list)
        hdg_mean = hdg_fin/len(connecting_rd_list)
        
        small_list = [j_id, incoming_road_list[i], connecting_rd_list, x_mean, y_mean, hdg_mean]
        big_list.append(small_list)
        
    print (big_list)
    return big_list
            
            
    
```

```python
# function to get mean junction coordinate
def get_mean_junction_coords(j_id):
    list_inc_road_info = get_coord_of_inc_road_to_junction(j_id)
    x_mean = 0
    y_mean = 0
    z_mean = 0
    for i in range(len(list_inc_road_info)):
        x_mean = x_mean+list_inc_road_info[i][3]
        y_mean = y_mean+list_inc_road_info[i][4]
        z_mean = z_mean+list_inc_road_info[i][5]
    x_mean_fin = x_mean/len(list_inc_road_info)
    y_mean_fin = y_mean/len(list_inc_road_info)
    z_mean_fin = z_mean/len(list_inc_road_info)
    mean_coords = [x_mean_fin, y_mean_fin, z_mean_fin]
    print (mean_coords)
    return mean_coords

get_mean_junction_coords(37)
        
```

```python
#function to convert opendrive coordinates to lat/long
from pyproj import Proj, transform

# from pyproj import Transformer

def opendrive_coords_to_lat_long(x,y):
    # offset of the opendrive file
    # xodr_x0=-639371.48
    # xodr_y0=-3848823.33    
    xodr_x0 = -667733.87
    xodr_y0=-3878705.38


    # initialize projection objects
    wgs84Proj = Proj('EPSG:4326')
    xodrProj = Proj('EPSG:32616')
    
#     transformer = Transformer.from_crs("EPSG:4326", "EPSG:3857")
#     transformer.transform(12, 12)

    
    # do the projection
    wgs84_x0, wgs84_y0 = transform(xodrProj, wgs84Proj, x-xodr_x0, y-xodr_y0)
    print('xodr/sumo coordinates {},{}, wgs84 coordinates {},{}'.format(x, y, wgs84_x0, wgs84_y0))
    fin_list = [x, y, wgs84_x0, wgs84_y0]
    return fin_list
```
