```python
import random

# Initialize a global counter for VISSIM runs
vissim_run_count = 0
vissim_run_times = []

# Objective function for simulated annealing
def objective_function(params, tt_measurement_ids, observed_tt):
    global vissim_run_count, vissim_run_times

    # Set the simulation resolution
    resolution = 10
    Vissim.Simulation.SetAttValue("SimRes", resolution)
    
    # Set the random seed for reproducibility
    vissim_seed = 21
    Vissim.Simulation.SetAttValue("RandSeed", vissim_seed)

    # Start the simulation and measure time taken
    start_vissim_time = time.time()

    # Update driver behavior parameters based on the params input
    set_driver_behavior_parameters(Vissim, params)

    # Run the simulation
    Vissim.Simulation.RunContinuous()

    end_vissim_time = time.time()
    vissim_run_duration = end_vissim_time - start_vissim_time
    vissim_run_times.append(vissim_run_duration)
    vissim_run_count += 1

    # Collect simulated travel time data
    simulated_tt = get_travel_time(Vissim, tt_measurement_ids)
    print (simulated_tt[0], simulated_tt[1])

    # Calculate the mean absolute error (MAE)
    fitness_mae = calc_mae(observed_tt, simulated_tt, len(tt_measurement_ids))

    return fitness_mae

# Simulated Annealing function
def simulated_annealing(objective_function, tt_measurement_ids, observed_tt, initial_params, initial_temp, cooling_rate, param_ranges, max_iter):
    current_params = initial_params.copy()
    current_cost = objective_function(current_params, tt_measurement_ids, observed_tt)
    
    # Print the initial solution and its fitness
    print(f"Initial Solution: {current_params}, Initial Fitness (MAE): {current_cost}")
    
    best_solution = current_params.copy()
    best_cost = current_cost
    
    history = [(vissim_run_count, current_cost)]
    temp = initial_temp
    iter = 0

    while temp > 0.001 and iter < max_iter:
        print(f"\nIteration {iter + 1}")

        new_params = current_params.copy()

        # Randomly alter parameters within their defined ranges
        for i, param_range in enumerate(param_ranges):
            new_params[i] = random.uniform(param_range[0], param_range[1])

        new_cost = objective_function(new_params, tt_measurement_ids, observed_tt)
        cost_diff = new_cost - current_cost
        
        # Print the solution being investigated and its fitness
        print(f"Investigating Solution: {new_params}, Fitness (MAE): {new_cost}")
        
        # Calculate the acceptance probability
        if new_cost < current_cost:
            accept_prob = 1.0
        else:
            accept_prob = np.exp((current_cost - new_cost) / temp)

        # Decide whether to accept the new solution
        if accept_prob > random.random():
            current_params = new_params.copy()
            current_cost = new_cost
            
            # Update the best solution if the new one is better
            if new_cost < best_cost:
                best_solution = new_params.copy()
                best_cost = new_cost

        # Track the fitness over iterations
        history.append((vissim_run_count, current_cost))

        # Decrease the temperature
        temp *= cooling_rate
        iter += 1

    # Print the final solution and its fitness
    print(f"\nFinal Solution: {best_solution}, Final Fitness (MAE): {best_cost}")
    
    return best_solution, best_cost, history


```

```python
def calculate_geh(observed, simulated):
    """Calculates the GEH statistic for a single pair of observed and simulated values."""
    if observed + simulated == 0:
        return 0  # Avoid division by zero
    return np.sqrt(2 * ((observed - simulated) ** 2) / (observed + simulated))

def calculate_geh_dict(observed_dict, simulated_dict):
    """Calculates GEH for each direction at each intersection and returns a dictionary of GEH values."""
    geh_values = {}

    for intersection in observed_dict:
        # if intersection in simulated_dict:
        observed_counts = observed_dict[intersection]
        simulated_counts = simulated_dict[intersection]
        
        if len(observed_counts) != len(simulated_counts):
            print(f"Warning: Mismatch in data length at {intersection}.")
            continue
        
        geh_list = []
        for obs, sim in zip(observed_counts, simulated_counts):
            geh = calculate_geh(obs, sim)
            geh_list.append(geh)
        
        geh_values[intersection] = geh_list
        # else:
            # print(f"Warning: Intersection {intersection} not found in simulated data.")
    
    return geh_values
```

```python
def calculate_geh(observed, simulated):
    """Calculates the GEH statistic for a single pair of observed and simulated values."""
    if observed + simulated == 0:
        return 0  # Avoid division by zero
    return np.sqrt(2 * ((observed - simulated) ** 2) / (observed + simulated))

def calculate_geh_dict(observed_dict, simulated_dict):
    """Calculates GEH for each direction at each intersection and returns a dictionary of GEH values."""
    geh_values = {}

    for intersection in observed_dict:
        # if intersection in simulated_dict:
        observed_counts = observed_dict[intersection]
        simulated_counts = simulated_dict[intersection]
        
        if len(observed_counts) != len(simulated_counts):
            print(f"Warning: Mismatch in data length at {intersection}.")
            continue
        
        geh_list = []
        for obs, sim in zip(observed_counts, simulated_counts):
            geh = calculate_geh(obs, sim)
            geh_list.append(geh)
        
        geh_values[intersection] = geh_list
        # else:
            # print(f"Warning: Intersection {intersection} not found in simulated data.")
    
    return geh_values
```

```python
import random
import numpy as np
import matplotlib.pyplot as plt

# Initialize global variables for VISSIM runs (simulated for this example)
vissim_run_count = 0
vissim_run_times = []

# Simulated Annealing Parameters
initial_temperature = 1000
cooling_rate = 0.99  # Slightly faster cooling rate
min_temperature = 0.1
max_iterations = 2000  # Increase iterations to allow more exploration
num_trials = 1  # Number of different initial solutions to try

# Function to generate a neighboring solution
def get_neighbor(solution, gene_space):
    neighbor = solution.copy()
    gene_idx = random.randint(0, len(solution) - 1)
    
    if gene_idx < 12:  # Turn ratios
        neighbor[gene_idx] = random.choice(np.arange(gene_space[gene_idx]['low'], gene_space[gene_idx]['high'] + gene_space[gene_idx]['step'], gene_space[gene_idx]['step']))
    else:  # Vehicle inputs
        neighbor[gene_idx] = random.choice(np.arange(gene_space[gene_idx]['low'], gene_space[gene_idx]['high'] + gene_space[gene_idx]['step'], gene_space[gene_idx]['step']))
    
    return neighbor

def simulated_annealing(initial_solution):
    global vissim_run_count
    current_solution = initial_solution
    mean_geh, geh_below_5_percent, current_fitness = objective_function(current_solution)
    print("Initial Fitness:", current_fitness)
    best_solution = current_solution
    best_fitness = current_fitness
    best_mean_geh = mean_geh
    best_geh_below_5_percent = geh_below_5_percent

    temperature = initial_temperature
    fitness_history = [current_fitness]
    mean_geh_history = [mean_geh]
    geh_below_5_percent_history = [geh_below_5_percent]

    while temperature > min_temperature and vissim_run_count < max_iterations:
        neighbor_solution = get_neighbor(current_solution, gene_space)
        mean_geh, geh_below_5_percent, neighbor_fitness = objective_function(neighbor_solution)
        
        # Decide whether to move to the neighbor solution
        if neighbor_fitness > current_fitness or random.uniform(0, 1) < np.exp((neighbor_fitness - current_fitness) / temperature):
            current_solution = neighbor_solution
            current_fitness = neighbor_fitness
            best_mean_geh = mean_geh
            best_geh_below_5_percent = geh_below_5_percent
        
        # Update the best solution
        if current_fitness > best_fitness:
            best_solution = current_solution
            best_fitness = current_fitness
        
        fitness_history.append(current_fitness)
        mean_geh_history.append(mean_geh)
        geh_below_5_percent_history.append(geh_below_5_percent)
        
        # Cool down
        temperature *= cooling_rate
        vissim_run_count += 1
    
    return best_solution, best_fitness, fitness_history, mean_geh_history, geh_below_5_percent_history

# Gene space with step changes
gene_space = [{'low': 0, 'high': 1, 'step': 0.01}] * 12 + [{'low': 50, 'high': 200, 'step': 5}] * 4


```
